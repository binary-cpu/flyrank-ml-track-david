## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Chosen Lane: Content Refresh Prioritization

ML Task Type: Binary Classification (predicting whether a page is declining) with a downstream Scoring & Ranking layer.

Explanation:
I am framing this problem as a Binary Classification task to predict whether a given search engine page is experiencing a significant performance decline (is_declining_label = 1) versus remaining stable or growing (0).

The predicted probability output from the classification model is then transformed into a continuous Priority / Refresh Score, which ranks pages from highest to lowest need for an SEO content update.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target / Proxy Variable: is_declining_label

Target Construction:
The target is a binary proxy label constructed by comparing recent page performance (Period 1 / p1) against historical baseline performance (Period 0 / p0). A page is marked as 1 (declining) if it experiences a substantial drop in organic clicks, impressions, or click-through rate (CTR) across comparison periods.

Data Leakage Constraints (Critical Guardrail):
Features derived directly from period-over-period trend calculations or target logic such as trend_direction, trend_pct, health_score, or needs_ctr_fix—must be excluded from training features. Including them creates direct target leakage, giving the model artificially perfect accuracy that fails in production.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Metric: Precision@K (specifically Precision@50 and Precision@100).Secondary Metrics: ROC-AUC and PR-AUC.Why Precision@K Matches Business Reality: An SEO content team has fixed bandwidth they can only manually inspect, rewrite, or update a limited number of pages per month (e.g., $K = 50$ or $100$ pages).Precision@K measures the proportion of true declining pages contained within the top $K$ recommendations generated by the model. Minimizing false positives in the top $K$ list ensures the team spends time updating pages that genuinely need intervention.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*


1.   Failure of Static Rule Thresholds:
Fixed heuristics (e.g., "Flag if clicks fall by more than 25%") apply rigid, uniform logic. They create high false-positive rates on low-traffic pages with random statistical noise, while missing high-volume pages where a small percentage drop represents a massive loss in traffic.

2.   Complex Feature Interactions:
Machine Learning algorithms (such as Random Forest or Gradient Boosted Trees) evaluate multiple features simultaneously ranking shifts, impression decay, CTR variances, and historical trends—without requiring manual threshold adjustments for every unique account type.
